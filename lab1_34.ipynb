**Сумма цифр факториала**

**Выполнил:** Хорошавин Б.И., группа ИНБс-4301

```python
def sum_digits_factorial(n):
    #Вычисляем факториал
    factorial = 1
    for i in range(1, n + 1):
        factorial = factorial * i

    #Превращаем число в строку
    factorial_str = str(factorial)

    #Считаем сумму цифр
    digit_sum = 0
    for digit in factorial_str:
        digit_sum = digit_sum + int(digit) # превращаем символ обратно в число

    return digit_sum
```

```python
# Тест 1: Пример из задания (10!)
print("10! = 3628800, сумма цифр:", sum_digits_factorial(10))

# Тест 2: Еще один пример (5!)
print("5! = 120, сумма цифр:", sum_digits_factorial(5))

# Тест 3: Граничный случай (0!)
print("0! = 1, сумма цифр:", sum_digits_factorial(0))
```
